```python
import pandas as pd 
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
df=pd.read_csv('Lung Cancer.csv')
df
```

<div style="background-color:#E67E22; color:white; padding:30px; border-radius:15px; text-align:center;">
    <h2 style="font-size:50px; margin:0;">1️⃣ Data Cleaning & Preprocessing 🧹</h2>
</div>

<br>

Before any analysis, we must **prepare the dataset** properly:  

### <span style="color:#3498DB;">Steps Taken:</span>
1. **Handling Missing Values:** Checked all columns for nulls; none found, dataset is complete.  
2. **Date Columns Conversion:** <span style="color:#F39C12;">`diagnosis_date`</span> and <span style="color:#F39C12;">`end_treatment_date`</span> converted to datetime objects for calculation.  
3. **Feature Engineering:**  
   - <span style="color:#F39C12;">`age_group`</span> categorized into: 0-30, 31-50, 51-70, 71+  
   - <span style="color:#F39C12;">`bmi_category`</span> categorized into: Underweight, Normal, Overweight, Obese  
   - <span style="color:#F39C12;">`treatment_duration_days`</span> = (<span style="color:#F39C12;">`end_treatment_date`</span> - <span style="color:#F39C12;">`diagnosis_date`</span>).days  
4. **Categorical Encoding:** For analysis and pivot tables (<span style="color:#F39C12;">`gender`</span>, <span style="color:#F39C12;">`smoking_status`</span>, <span style="color:#F39C12;">`treatment_type`</span>, <span style="color:#F39C12;">`family_history`</span>)  
5. **Risk Category Creation:** Combining factors like BMI, Age, Cancer Stage, and Comorbidities to define Low, Medium, High Risk  

> This preprocessing ensures that the dataset is **ready for deep analysis** and all insights are accurate and meaningful.


```python
df['diagnosis_date'] = pd.to_datetime(df['diagnosis_date'])
df['end_treatment_date'] = pd.to_datetime(df['end_treatment_date'])
```

```python
df['treatment_duration_days'] = (df['end_treatment_date'] - df['diagnosis_date']).dt.days
```

```python
df.head(5)
```

```python
df['treatment_duration_days'].describe()
```

```python
bins = [0, 30, 50, 70, 105]
labels = ['0-30', '31-50', '51-70', '71+']
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels)
```

```python
df.head(5)
```

```python
def bmi_category(bmi):
    if bmi < 18.5:
        return 'Underweight'
    elif bmi < 25:
        return 'Normal'
    elif bmi < 30:
        return 'Overweight'
    else:
        return 'Obese'

df['bmi_category'] = df['bmi'].apply(bmi_category)
```

```python
df[['age', 'age_group', 'bmi', 'bmi_category', 'treatment_duration_days', 'survived']].head()
```

<div style="background-color:#8E44AD; color:white; padding:30px; border-radius:15px; text-align:center;">
    <h2 style="font-size:50px; margin:0;">2️⃣ Survival Analysis by Single Category 📊</h2>
</div>

<br>

We start by analyzing **survival rates individually** by key categories to understand broad patterns.

### <span style="color:#3498DB;">Categories Analyzed:</span>
- **Cancer Stage:** Stage I, II, III, IV  
- **Age Group:** 0-30, 31-50, 51-70, 71+  
- **Gender:** Male, Female  
- **BMI Category:** Underweight, Normal, Overweight, Obese  
- **Smoking Status:** Never Smoked, Current Smoker, Passive Smoker, Former Smoker  
- **Family History:** Yes, No  
- **Treatment Type:** Surgery, Chemotherapy, Radiation, Combined  
- **Risk Category:** Low, Medium, High  

### <span style="color:#3498DB;">Insights Example:</span>
- Stage IV patients have the **lowest survival rates (~22%)**  
- Younger patients (0-30) show **higher survival even in advanced stages**  
- Surgery patients slightly outperform Chemotherapy in survival  
- Risk Category confirms **High Risk patients consistently have lower survival**


```python
sns.countplot(x='cancer_stage', hue='survived', data=df)
plt.show()

sns.countplot(x='age_group', hue='survived', data=df)
plt.show()

sns.countplot(x='bmi_category', hue='survived', data=df)
plt.show()

sns.countplot(x='treatment_type', hue='survived', data=df)
plt.show()
```

```python
numeric_cols = ['age', 'bmi', 'cholesterol_level', 'treatment_duration_days']
correlations = df[numeric_cols + ['survived']].corr()
print(correlations['survived'].sort_values(ascending=False))
```

```python
categorical_cols = ['cancer_stage', 'gender', 'smoking_status', 'family_history', 
                    'treatment_type', 'bmi_category', 'age_group']

for col in categorical_cols:
    survival_rate = df.groupby(col)['survived'].mean().sort_values(ascending=False)
    print(f"\nSurvival rate by {col}:\n", survival_rate)
```

```python
sns.boxplot(x='survived', y='treatment_duration_days', data=df)
plt.show()
```

<div style="background-color:#2980B9; color:white; padding:30px; border-radius:15px; text-align:center;">
    <h2 style="font-size:50px; margin:0;">3️⃣ Multi-Dimensional Analysis 🔄</h2>
</div>

<br>

To extract **deeper insights**, we combined multiple dimensions using pivot tables and group-by analyses.

### <span style="color:#3498DB;">Combinations Analyzed:</span>
- Cancer Stage + Treatment + Age Group  
- Cancer Stage + BMI + Smoking Status  
- Treatment Type + Family History + Age Group  
- Cancer Stage + Treatment + BMI + Age Group  

### <span style="color:#3498DB;">Key Findings:</span>
- Younger patients in advanced stages still survive better with Surgery.  
- High Risk patients consistently have lower survival, regardless of treatment type.  
- Certain BMI & Smoking Status combinations reveal hidden survival stratifications.  
- Treatment effectiveness varies with age and cancer stage.


```python

pivot1 = df.pivot_table(index=['cancer_stage', 'treatment_type', 'age_group'],
                        values='survived', aggfunc='mean').sort_values(by='survived', ascending=False)
print("Survival rate by Stage + Treatment + Age Group:\n", pivot1.head(20))


pivot2 = df.pivot_table(index=['cancer_stage', 'bmi_category', 'smoking_status'],
                        values='survived', aggfunc='mean').sort_values(by='survived', ascending=False)
print("\nSurvival rate by Stage + BMI + Smoking Status:\n", pivot2.head(20))


pivot3 = df.pivot_table(index=['treatment_type', 'family_history', 'age_group'],
                        values='survived', aggfunc='mean').sort_values(by='survived', ascending=False)
print("\nSurvival rate by Treatment + Family History + Age Group:\n", pivot3.head(20))

p
pivot4 = df.pivot_table(index=['cancer_stage', 'treatment_type', 'bmi_category', 'age_group'],
                        values='survived', aggfunc='mean').sort_values(by='survived', ascending=False)
print("\nSurvival rate by Stage + Treatment + BMI + Age Group:\n", pivot4.head(20))
```

```python
stage_score = {
    'Stage I': 1,
    'Stage II': 2,
    'Stage III': 3,
    'Stage IV': 4
}

age_score = {
    '0-30': 1,
    '31-50': 2,
    '51-70': 3,
    '71+': 4
}

bmi_score = {
    'Normal': 1,
    'Overweight': 2,
    'Obese': 3,
    'Underweight': 4
}

treatment_score = {
    'Surgery': 1,
    'Radiation': 2,
    'Combined': 3,
    'Chemotherapy': 4
}

smoking_score = {
    'Never Smoked': 1,
    'Passive Smoker': 2,
    'Former Smoker': 3,
    'Current Smoker': 4
}

family_score = {
    'No': 1,
    'Yes': 2
}


df['risk_score'] = (
    df['cancer_stage'].map(stage_score).astype(int) +
    df['age_group'].map(age_score).astype(int) +
    df['bmi_category'].map(bmi_score).astype(int) +
    df['treatment_type'].map(treatment_score).astype(int) +
    df['smoking_status'].map(smoking_score).astype(int) +
    df['family_history'].map(family_score).astype(int)
)


#  نصنف المرضى حسب Risk Score
# Low Risk: 6-9, Medium Risk: 10-14, High Risk: 15+
def classify_risk(score):
    if score <= 9:
        return 'Low Risk'
    elif score <= 14:
        return 'Medium Risk'
    else:
        return 'High Risk'

df['risk_category'] = df['risk_score'].apply(classify_risk)


risk_distribution = df['risk_category'].value_counts(normalize=True) * 100
print("Risk Category Distribution (%):\n", risk_distribution)


survival_by_risk = df.groupby('risk_category')['survived'].mean()
print("\nSurvival rate by Risk Category:\n", survival_by_risk)
```

```python
#  Heatmap: Survival rate by Stage + Treatment + Age Group
pivot_heat = df.pivot_table(index='cancer_stage', columns=['treatment_type', 'age_group'],
                            values='survived', aggfunc='mean')
plt.figure(figsize=(12,6))
sns.heatmap(pivot_heat, annot=True, fmt=".2f", cmap='YlGnBu')
plt.title('Survival Rate Heatmap: Stage x Treatment x Age Group')
plt.show()
```

```python
#  Boxplot: Treatment Duration vs Risk Category
plt.figure(figsize=(8,5))
sns.boxplot(x='risk_category', y='treatment_duration_days', data=df)
plt.title('Treatment Duration by Risk Category')
plt.show()
```
